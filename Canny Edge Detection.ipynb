```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python

img = cv2.imread(r"C:\Users\admin\DIPT\workshop3\SEC.jpg")
```

```python
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
img_blur = cv2.GaussianBlur(img_gray, (5,5), 0)
```

```python
edges = cv2.Canny(img_gray, threshold1 = 180, threshold2 = 200)
```

```python

plt.figure(figsize=(18,6))
```

```python
plt.subplot(1,4,1)
plt.imshow(img[:,:,::-1])
plt.axis('off')
plt.title('Original')
```

```python
plt.subplot(1,4,2)
plt.imshow(img_gray, cmap='gray')
plt.axis('off')
plt.title('Grayscale')
```

```python
plt.subplot(1,4,3)
plt.imshow(img_blur, cmap='gray')
plt.axis('off')
plt.title('Blurred')
```

```python
plt.subplot(1,4,4)
plt.imshow(edges, cmap='gray')
plt.axis('off')
plt.title('Canny Edge Map')
```
